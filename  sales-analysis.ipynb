```python
import pandas as pd
import random

items = {"Crepe": 60, "Waffle": 80, "Pizza": 120, "Juice": 30}
rows = []

for day in range(1, 31):
    for _ in range(10):
        item = random.choice(list(items))
        qty = random.randint(1, 4)
        rows.append([f"2026-11-{day:02d}", item, qty, items[item] * qty])

df = pd.DataFrame(rows, columns=["date", "item", "qty", "total"])
print(df.head())
```

```python
print("Total sales:", df["total"].sum())

by_item = df.groupby("item")["total"].sum().sort_values(ascending=False)
print(by_item)

print("Best seller:", by_item.idxmax())
```

```python
import matplotlib.pyplot as plt

by_item.plot(kind="bar", color="steelblue")
plt.title("Sales by Item")
plt.xlabel("Item")
plt.ylabel("Total Sales (EGP)")
plt.xticks(rotation=0)
plt.show()
```
